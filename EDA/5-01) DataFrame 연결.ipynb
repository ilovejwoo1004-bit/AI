# DataFrame Merge

서로 다른 두 개의 DataFrame을 하나로 합치는 작업은 두 가지 방식으로 구분할 수 있음.

 * concatenating(연결) : 단순히 하나의 DataFrame에 다른 DataFrame을 연속적으로 붙이는 방식

 * merging(병합) : 두 DataFrame에 공통적으로 포함되어 있는 하나의 열을 기준으로 삼아, 해당 열의 값이 동일한 두 개의 행들을 하나의 행으로 합치는 방식


## 1. Merging DataFrame

**열 기준 Merging**

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({'key': list('bbacaab'),
                   'data': range(7)})
df2 = pd.DataFrame({'key': list('abd'),
                   'data2': range(3)})
```

```python
df1
```

```python
df2
```

```python
# key 열을 기준으로 하여 두 DataFrame에 대한 merging을 수행
# == db의 inner join과 같은 역할
pd.merge(df1, df2, on ='key') 
```

```python
# how 인자의 값을 outer로 하면 다른 조합 방식에 따라 새로운 행들을 생성함
# == db의 outer join과 같은 역할
pd.merge(df1, df2, on='key', how='outer')
```

```python
# left(df1) 기준으로 merge
# == left join 같은 기능
pd.merge(df1, df2, on='key', how ='left') 
```

```python
# right(df2) 기준으로 merge
# == right join 같은 기능
pd.merge(df1, df2, on='key', how ='right') 
```

```python
# merge하고자 하는 기준 열의 이름이 다른 경우의 merge 방법

df3 = pd.DataFrame({'lkey': list('bbacaab'),
                    'data1': range(7)})
df4 = pd.DataFrame({'rkey': list('abd'),
                   'data2': range(3)})
```

```python
df3
```

```python
df4
```

```python
pd.merge(df3, df4, left_on='lkey', right_on='rkey')
```

**인덱스 기준 merging**

```python
left1 = pd.DataFrame({'key': ['a','b','a','a','b','c'],
                     'value': range(6)})
right1 = pd.DataFrame({'group_val':[3.5, 7]},
                      index=['a','b'])
```

```python
left1
```

```python
right1
```

```python
# left1에서 'key'열을, right1에서는 인덱스를 기준으로 merging을 수행
pd.merge(left1, right1, left_on='key', right_index=True, how='outer')
```

```python
#두 DataFrame 모두 인덱스 기준으로 merging을 수행하고자 할 경우

left2 = pd.DataFrame([[1.,2.],[3.,4.],[5.,6.]],
                    index=['a','c','e'], columns=['Seoul','Incheon'])
right2 = pd.DataFrame([[7.,8.],[9.,10.],[11.,12.],[13.,14.]],
                     index=['b','c','d','e'], columns=['Daegu', 'Ulsan'])
```

```python
left2
```

```python
right2
```

```python
pd.merge(left2, right2, how='outer', left_index=True, right_index=True)
```

** Concatenating DataFrame**

*: Concatenating은 쉽게 말해서 하나의 DataFrame에 다른 DataFrame을 행 방향 또는 열 방향으로 단순 연결하는 것을 의미*

```python
# Series data concatenating

s1 = pd.Series([0,1], index=['a', 'b'])
s2 = pd.Series([2,3,4], index=['c','d','e'])
s3 = pd.Series([5,6], index=['f','g'])
```

```python
s1
```

```python
s2
```

```python
s3
```

```python
pd.concat([s1,s2,s3])
```

```python
# axis=1 인자를 명시하면 Series가 2차원적으로 연결
pd.concat([s1,s2,s3], axis=1)
```

```python
s4 = pd.concat([(s1*5), s3])
```

```python
pd.concat([s1,s4], axis=1)
```

```python
# concatenating 진행시, key값으로 컬럼을 명시할 수 있음
pd.concat([s1,s2,s3], axis=1, keys=['one','two','three'])
```

```python
# DataFrame Concatenating

df1 = pd.DataFrame(np.arange(6).reshape(3,2),
                  index=['a','b','c'], columns=['one', 'two'])
df2 = pd.DataFrame(5 + np.arange(4).reshape(2,2),
                  index=['a','c'], columns=['three','four'])
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2], axis=1)
```

```python
# 행방향으로 concatenating을 수행할 때 기존 dataframe의 인덱스가 유지되었는데,
# ignore_index=True 인자를 넣어주면 기존의 인덱스를 무시하고 새로운 인덱스를 붙여줌

df3 = pd.DataFrame(np.random.randn(3, 4), columns=['a', 'b', 'c', 'd'])
df4 = pd.DataFrame(np.random.randn(2, 3), columns=['b', 'd', 'a'])
```

```python
df3
```

```python
df4
```

```python
pd.concat([df3, df4], ignore_index=True)
```

```python
# ignore_index=True 를 안줬을 경우
pd.concat([df3, df4])
```

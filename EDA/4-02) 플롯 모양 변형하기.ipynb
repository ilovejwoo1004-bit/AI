# 플롯 모양 변형하기
 
### Figure, subplots 및 axes
 
matplotlib에서는 'figure(피겨)'라는 그림 단위를 사용하여, 이 안에 한 개 혹은 복수 개의 플롯을 그리고 관리할 수 있도록 하는 기능을 지원한다. 이 때, figure 안에 들어가는 플롯 공간 하나를 'subplot(서브플롯)'이라고 부른다.
 

 
### matplotlib에서 사용 가능한 주요 color, marker, linestyle 값
 
matplotlib에서 사용 가능한 주요 color 값
 

  값  :  색상


- "b" : blue 
- "g" : green 
- "r" : red 
- "c" : cyan 
- "m" : magenta 
- "y" : yellow 
- "k" : black 
- "w" : white 
 
matplotlib에서 사용 가능한 주요 marker 값
 

  값 :  마킹


- "." : point 
- "," : pixel 
- "o" : circle 
- "v" : triangle_down 
- "^" : triangle_up 
- "<" : triangle_left 
- ">" : triangle_right 
- "8" : octagon 
- "s" : square 
- "p" : pentagon 
- "*" : star 
- "h" : hexagon 
- "+" : plus 
- "x" : x 
- "D" : diamond 
 
matplotlib에서 사용 가능한 주요 linestyle 값
 

  값  :  라인 스타일


- "-" : solid line 
- "--" : dashed line 
- "-." : dash-dotted line 
- ":" : dotted line 
- "None" : draw nothing

```python
data = pd.Series(np.random.randn(16), index = list('abcdefghijklmnop'))
```

```python
# 수직방향의 바플롯
data.plot(kind = 'bar', ax=axes[0], color='k', alpha=0.7)   # alpha : 투명도
```

```python
# 수평방향의 바플롯
data.plot(kind = 'barh', ax=axes[1], color='g', alpha=0.3)
```

```python
# 범례, 눈금 수정
fig = plt.figure()
```

```python
ax = fig.add_subplot(1,1,1)
```

```python
ax.plot(np.random.randn(1000).cumsum())
#1000개의 랜덤값을 만들어 이들의 누적합을 더한 값
```

```python
#수평축 변경
ax.set_xticks([0,250,500,750,1000])
```

```python
labels = ax.set_xticklabels(['one','two','three','four','five'],
                           rotation=30, fontsize='small')   # 문자레이블의 기울기, 글씨크기 지정
```

```python
# 수직축 변경
# --> set_yticks와 set_yticklabels 등을 이용하여 변경 가능
```

```python
# 제목 입력
ax.set_title('random walk plot')
```

```python
#수평축 이름 지정
ax.set_xlabel('stages')
#수직축 이름 지정
ax.set_ylabel('values')
```

```python
# 범례표시
fig = plt.figure()
ax = fig.add_subplot(1,1,1)
```

```python
ax.plot(np.random.randn(1000).cumsum(), 'k', label="one")
ax.plot(np.random.randn(1000).cumsum(), "b--", label="two")
ax.plot(np.random.randn(1000).cumsum(), "r.", label="three")
```

```python
ax.legend(loc='best')   #loc='best' 최적의 위치에 범례를 위치시킨다
```

```python
ax.get_xlim()
```

```python
#수평축 단위 변경
ax.set_xlim([100, 900])
```

```python
#수직축 단위 변경
ax.set_ylim([-100, 100])
```

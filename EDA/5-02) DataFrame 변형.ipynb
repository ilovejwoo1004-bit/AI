**데이터 매핑하기 **

```python
df2 = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon', 'Pastrami', 
                             'corned beef', 'Bacon', 'pastrami', 'honey ham',
                             'nova lox'],
                    'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
```

```python
df2
```

생성된 df2의 "food" 열에 몇 가지 값이 들어가 있는데, 이 "food" 열의 값들을 새로운 값으로 매핑하는 새로운 딕셔너리를 정의.

```python
meat_to_animal = { 
    'bacon': 'pig', 
    'pulled pork': 'pig', 
    'pastrami': 'cow', 
    'corned beef': 'cow', 
    'honey ham': 'pig', 
    'nova lox': 'salmon'
}
```

이 meat_to_animal 딕셔너리에 명시된 규칙에 의거하여, "food" 열의 값들을 "pig", "cow", "salmon" 중 하나로 치환한 뒤 이들을 갖는 새로운 "animal" 열을 다음과 같이 추가할 수 있다.

```python
df2["animal"] = df2["food"].apply(lambda x: meat_to_animal[x.lower()])
df2
```

: df2.apply() 함수를 사용하여, 기존 "food" 열에 포함된 각 문자열 값에 str.lower() 함수를 먼저 적용하여 모두 소문자로 변환한 뒤 이를 키로 갖는 meat_to_animal에서의 값을 반환하는 코드.

** Categories(카테고리) 자료형 **
 
어떤 데이터셋에서 유한 가지 값들 중 하나를 가지는 열이 존재하는 경우가 많다. 예를 들어 고객의 성별을 나타내는 열에서 '남성'이면 "male", '여성'이면 "female" 값을 가지는 것이 대표적인 경우. 이러한 유형의 데이터를 '카테고리(category, 범주형)' 데이터라고 함.
 
이런 카테고리 데이터를 관리하기 위해, pandas에서는 'Categories'라는 특별한 자료형을 제공. Categories 자료형을 활용하면, 단순히 문자열 자료형으로 저장되어 있던 데이터를 대체함으로써 메모리 사용량을 줄일 수 있으며, 이에 대한 각종 통계 분석도 보다 간편하게 할 수 있다.
 
**Categories 열 생성하기**

```python
df4 = pd.DataFrame({'id':[1,2,3,4,5,6],
                   'raw_grade': ['a','b','b','a','a','e']})
```

```python
df4
```

df4의 "raw_grade" 열에는 'a', 'b', 'e' 등의 카테고리 값이 단순 문자열 형태로 저장되어 있다. 이를 Categories 자료형으로 치환한 새로운 "grade" 열을 다음과 같이 정의.

```python
df4['grade'] = df4['raw_grade'].astype('category')
df4['grade']
```

df4["grade"]를 확인해보면 'Categories (3, object): [a, b, e]'라고 표시되는 것을 확인할 수 있는데, 이는 해당 열이 Categories 자료형의 Series라는 것이며, 이것이 'a', 'b', 'e'의 3가지 카테고리 값을 가질 수 있다는 것을 나타낸다.
 

```python
# 현재 "grade" 열이 가질 수 있는 카테고리 값 확인

df4['grade'].cat.categories
```

```python
# 새로운 카테고리를 포함하는 리스트 입력
df4["grade"].cat.categories = ["very bad", "good", "very good"]
df4
```

한편 기존 카테고리 값의 종류를 늘이거나 줄이는 것도 가능함. df4["grade"].cat.set_categories() 함수를 사용하여, 기존 카테고리 값에서 추가되거나 삭제된 것들을 리스트 형태의 인자로 명시하면 된다.

```python
df4["grade"] = df4["grade"].cat.set_categories(["very bad", "bad", "medium", "good", "very good"])
```

Categories 자료형의 또 다른 특징은, 처음 정의할 때 부여한 순서에 따라 대소 관계를 가진다는 것. 다음과 같이 "grade" 열의 값을 기준으로 DataFrame을 정렬

```python
df4.sort_values(by='grade')
```

**숫자 데이터의 카테고리화**
 
고객 데이터셋을 관리할 때 '연령'을 나타내는 열이 숫자 데이터로 구성되어 있었다고 하자. 이를 '청소년' -> '청년' -> '장년' -> '중년' 등과 같이 일정한 나이 기준으로 구간을 잘라서 카테고리 데이터로 변환하여 관리하면 좀 더 편리해지는 경우가 있다. 
pandas에서는 숫자 데이터를 빠르고 간편하게 카테고리화하는 기능을 제공함.
 

```python
ages = [20, 22, 25, 27, 21, 23, 37, 31, 61, 45, 41, 32]
bins = [18, 25, 35, 60, 100]
```

ages 리스트에 카테고리화 하고자 하는 숫자 데이터를, bins 리스트에 각 구간을 구분하는 숫자값을 명시함. 즉, 이렇게 하면 18세 초과 25세 이하, 25세 초과 35세 이하 등과 같이 총 4개의 구간을 정의. 
pd.cut() 함수를 사용, ages와 bins를 인자로 입력함으로써 ages 리스트의 각 성분들을 카테고리화 할 수 있다.

```python
cats = pd.cut(ages, bins)
cats
```

cats를 확인해보면 기존의 ages 리스트의 각 성분이 4가지 카테고리 값 중 하나로 대체된 결과물을 확인할 수 있다. cats.categories를 확인하면, 이 과정에서 사용된 카테고리 값들만을 직접 확인할수도 있다. 또 cats.codes를 확인하면, 기존 ages 리스트의 각 성분에 (0부터 시작하는 인덱스 형태로) 몇 번째 카테고리 값이 부여되었는지를 확인할 수도 있다. 
 
만약 "(18, 25]"와 같은 이름이 별로 보기 좋지 않다면, pd.cut() 함수를 실행할 때 labels 인자의 값을 입력함으로써 각 카테고리의 이름을 직접 명시할 수도 있다.


```python
cats.categories
```

```python
cats.codes
```

```python
# 카테로리를 문자로 표현
group_names = ["Youth", "YoungAdult", "MiddleAged", "Senior"]
pd.cut(ages, bins, labels=group_names)
```

만약 구간을 구분하는 기준값을 명시하는 것이 귀찮다면, pd.cut() 함수를 호출할 때 몇 개의 구간으로 나눌 것인지만 명시해주면 pandas에서 알아서 판단하여 길이가 일정한 구간을 정의하고 각 성분을 카테고리화한다. 이 때 precision 인자는 구간을 구분하는 기준값을 결정할 때 소수 몇 번째 자리까지 고려할지 결정.

```python
data = np.random.rand(20)
pd.cut(data, 4, precision=2)
```

한편 pd.qcut()이라는 함수도 있는데, 이는 지정한 갯수만큼의 구간을 정의하되, 분위수를 구분값으로 한다.

```python
data2 = np.random.randn(1000)
cats = pd.qcut(data2, 4)
cats
```
